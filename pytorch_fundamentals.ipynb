### Matrix multiplication

Two main ways of performing multiplication in neural networks and deep learning:

1. Elementwise multiplication
2. Matrix multiplication (dot product)

There are 2 main rules that performing matrix multiplication needs to satisfy:
1. The **inner Dimensions** must match:
* `(3,2) @ (3,2)` wont work
* `(2,3) @ (3,2)` will work
* `(3,2) @ (2,3)` will work
2. The resulting matrix has the shape of the **outer dimensions**:
* `(2,3) @ (3,2) -> (2,2)`
* `(3,2) @ (2,3) -> (3,3)`

```python
# Element wise multiplication
print(f"tensor is: {tensor}")
print(tensor * tensor)
```

```python
# Matrix multiplication
torch.matmul(tensor, tensor)
```

## Reshaping, stacking, squeezing and unsqueezing tensors

* Reshapping - reshapes an input tensor into defined shape
* View - Return a view of an input tensor of certain shape but keep same memory as the original tensor
* Stacking - combine multiple tensor on top of each other
* Squeeze - removes all `1` dimension tensor
* Unsqueeze - add a `1` dimension to a target tensor
* Permute - Return a view of the input with dimesions permuted in a certain way

```python
# create a tensor
import torch

x = torch. arange(1., 10.)
x, x.shape
```

```python
# Add an extra dimension
x_reshaped = x.reshape(1, 9)
x_reshaped, x_reshaped.shape
```

```python
# Change the view
z = x.view(1,9)
z, z.shape
```

```python
# changing z changes x. Since view of a tensor shares same memory with original tensor
z[:,0] = 5
z, x
```

```python
# Stack tensor on top ofeach other
x_stacked = torch.stack([x,x,x,x], dim =1)
x_stacked
```

```python
# torch.squeeze() removes all single dimensions
x_squeeze = torch.squeeze(x)
x_squeeze, x_squeeze.shape
```

```python
#Unsqueeze the tensor
x_unsqueeze = x.unsqueeze(dim=1)
x_unsqueeze
```

```python
# torch.permute() - rearrange the dimensions of a target tensor in specified order
x_original = torch.rand(size=(224,224,3))
x_original

# permute original tensorto rearrange axis
x_permute = x_original.permute(2,1,0)
x_permute
```

## Pytorch tensors and numpy

Numpy is a popular scientific python numerical computing library
and pytorch has functionality to interact with it

* Data is numpy, want in pytorch tensor -> `torch.from_numpy(ndarray)`
* Pytorch tensor -> Numpy -> `torch.Tensor.numpy()`

```python
# Numpy array to tensor
import numpy as np
import torch

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)
array, tensor
```

```python
# Tensor to numpy
tensor = torch.ones(7)
numpy_tensor = tensor.numpy()
tensor, numpy_tensor
```

## Reproducibility (trying to take random out of random)

In short how a neural network learns:
`start with random numvbers -> tensor operations -> update random numers to try and make them better representation of data ->again`

To reduce the randomness in neural networks and Pytorch comes the concept of a
**random seed**.

Essentially what random seed does is "flavour" the randomness


```python
import torch

#create 2 random tensortas

random_tensor_A = torch.rand(3,3)
random_tensor_B = torch.rand(3,3)

random_tensor_A, random_tensor_B
```

```python
# Lets make some random but reproducable tensors
import torch

# set the random seed
RANDOM_SEED = 42

torch.manual_seed(RANDOM_SEED)
random_tensor_c = torch.rand(3,4)

torch.manual_seed(RANDOM_SEED)
random_tensor_d = torch.rand(3,4)

random_tensor_c, random_tensor_d
```
